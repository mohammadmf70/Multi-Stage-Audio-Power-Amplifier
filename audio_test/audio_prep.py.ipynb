```python
import numpy as np
import soundfile as sf

input_file = r"input.wav"
output_file = r"audio.wav"

# خواندن فایل
data, samplerate = sf.read(input_file)

# ۱. تک‌کاناله کردن (Mono)
if len(data.shape) > 1:
    data = np.mean(data, axis=1)

# ۲. نرمال‌سازی دامنه (بین -0.8 تا 0.8 برای جلوگیری از Clipping)
max_val = np.max(np.abs(data))
if max_val > 0:
    data = (data / max_val) * 0.8

# ۳. ذخیره با انکودینگ کاملاً استاندارد قابل فهم برای LTspice
# نکته مهم: تعیین pcm_16 به همراه format='WAV'
sf.write(output_file, data, samplerate, subtype='PCM_16', format='WAV')

print("فایل صوتی با موفقیت آماده شد.")
```
